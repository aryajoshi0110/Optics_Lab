```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Part 1: Communication between Alice and Bob

data1 = pd.read_excel('Expt4_Team16(digital).xlsx', sheet_name='Sheet1')

Alice_base_choice = data1.iloc[2:,0].values  # Alice's base choice
Alice_bit_sent = data1.iloc[2:,4].values  # Alice's bit sent
Bob_base_choice = data1.iloc[2:,5].values  # Bob's base choice
Bob_bit_received = data1.iloc[2:,7].values  # Bob's bit received

print("Alice's base choice:", Alice_base_choice)
```

```python
Key = ""
for i in range(len(Alice_base_choice)):
    if Alice_base_choice[i] == Bob_base_choice[i]:
        Key += str(Alice_bit_sent[i])

print("Shared Key:", Key)
```

```python
# Encryption and Decryption
Example_messages = ["Ashok" , "Shivam" , "Neha" , "Arya" ]
Binary_messages = [''.join(format(ord(char), '08b') for char in message) for message in Example_messages]
Encrypted_messages, Decrypted_messages = [], []
# Converting message to binary and then encrypting using XOR with the key
for message in Example_messages:
    binary_message = ''.join(format(ord(char), '08b') for char in message)
    encrypted_message = ''.join(str(int(binary_message[i]) ^ int(Key[i % len(Key)])) for i in range(len(binary_message)))
    Encrypted_messages.append(encrypted_message)

# Decrypting the encrypted messages
for encrypted_message in Encrypted_messages:
    decrypted_message = ''.join(str(int(encrypted_message[i]) ^ int(Key[i % len(Key)])) for i in range(len(encrypted_message)))
    Decrypted_messages.append(decrypted_message)

print("Key:", Key)
print()

print("Binary Messages:")
for i, msg in enumerate(Example_messages):
    print(f"{msg}: {Binary_messages[i]}")
print()

print("Encrypted Messages:")
for i, msg in enumerate(Example_messages):
    print(f"{msg}: {Encrypted_messages[i]}")
print()

print("Decrypted Messages:")
for i, msg in enumerate(Example_messages):
    print(f"{msg}: {Decrypted_messages[i]}")
print()

```

```python
# Part 2: Eavesdropping by Eve
data2 = pd.read_excel('Expt4_Team16(digital).xlsx', sheet_name='Sheet2')

Alice_base_choice_eve = data2.iloc[2:,0].values  # Alice's base choice
Alice_bit_sent_eve = data2.iloc[2:,4].values  # Alice's bit sent
Eve_base_choice = data2.iloc[2:,5].values  # Eve's base choice
Eve_bit_received = data2.iloc[2:,7].values  # Eve's bit received
Bob_base_choice_eve = data2.iloc[2:,8].values  # Bob's base choice
Bob_bit_received_eve = data2.iloc[2:,10].values  # Bob's bit received
```

```python
# Efficiency Calculation

equal_bases = 0
successful_transmissions = 0

for i in range(len(Alice_base_choice_eve)):
    if Alice_base_choice_eve[i] == Bob_base_choice_eve[i]:
        equal_bases += 1
        if Alice_bit_sent_eve[i] == Bob_bit_received_eve[i]:
            successful_transmissions += 1

Error = (equal_bases - successful_transmissions) / equal_bases * 100 if equal_bases > 0 else 0
print(f"Total bits sent: {len(Alice_bit_sent_eve)}")
print(f"Bits where Alice and Bob used the same base: {equal_bases}")
print(f"Successful transmissions: {successful_transmissions}")
print(f"Error rate: {Error:.2f}%")
```
